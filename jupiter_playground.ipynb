# Aggregation

```python
# dfcontent['fieldname'].sum()  # summing up this column's value, ignoring Null/NaN
# dfcontent.sum() # summing up every column's value, make a table
```

```python
missing_english_title_filter = dfcontent['title_english'].isna() # create a filter for given column null value
```

```python
df_content_missing_english_title = dfcontent.loc[missing_english_title_filter, :]
```

```python
df_content_missing_english_title.head()
```

```python
dfcontent.loc[-missing_english_title_filter, :].head() # NAGATE the filter
```

```python
dfcontent['title_english'].isna().sum() # True=1/False=0 in python, this way it turn each row into True/False => 1/0, then sum up; typical use to check null value count for a column in DF
```

```python
dropped = dfcontent['title_english'][5000:6000].dropna(how='any')
```
